##map()

##map() is used when you want to perform the same operation on every element of an iterable

```python
#map(function, iterable)

numbers = [1, 2, 3, 4, 5]

def square(x):

    return x*x

result = map(square, numbers)

print(list(result))

# with lambda function

numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x * x, numbers)

print(list(result))

'''numbers
   ↓
[1, 2, 3, 4, 5]
   ↓
lambda x: x*x
   ↓
[1, 4, 9, 16, 25]'''
```

```python
numbers = [5, 10, 15, 20]

result = map(lambda x: x + 10, numbers)

print(list(result))

```

##filter()

##filter() is used when you want to select only certain elements from an iterable

```python
#filter(function, iterable)

numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, numbers)

print(list(result))

'''
Working
1 → 1 % 2 == 0 → False → ❌
2 → 2 % 2 == 0 → True  → ✅
3 → 3 % 2 == 0 → False → ❌
4 → 4 % 2 == 0 → True  → ✅
5 → 5 % 2 == 0 → False → ❌
6 → 6 % 2 == 0 → True  → ✅

Result → [2, 4, 6]'''
```

```python
numbers = [20, 55, 30, 80, 90, 10]

result = filter(lambda x: x > 50, numbers)

print(list(result))
```

```python
names = ["Aman", "Rahul", "Ankit", "Harsh", "Ajay"]

result = filter(lambda name: name.startswith("A"), names)

print(list(result))
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = filter(lambda x: x % 2 == 0, numbers)

squares = map(lambda x: x * x, even_numbers)

print(list(squares))


result = map(
    lambda x: x * x,
    filter(lambda x: x % 2 == 0, numbers)
)

print(list(result))
```

```python
a = [1, 2, 3, 4]
b = [10, 20, 30, 40]

result = map(lambda x, y: x + y, a, b)

print(list(result))
```
