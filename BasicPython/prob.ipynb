```python
import numpy as np

# Simulate 10,000 coin tosses
tosses = np.random.choice(
    ["Heads", "Tails"],
    size=10000
)

# Calculate the observed probability of heads
prob_heads = np.mean(tosses == "Heads")

print("Probability of heads:", prob_heads)
```

```python
import numpy as np

tosses = np.random.choice(["Heads", "Tails"], size=10000)

prob_heads = np.mean(tosses == "Heads")
prob_tails = np.mean(tosses == "Tails")

print("probability of Heads:", prob_heads, "probability of Tails:", prob_tails)
```

```python
import pandas as pd

df = pd.DataFrame({
    "discount": [
        True, True, True, True, True,
        False, False, False, False, False
    ],
    "purchased": [
        True, True, False, True, False,
        True, False, False, False, False
    ]
})

discount_customers = df[df["discount"] == True]

probability = discount_customers["purchased"].mean()

print("Purchase probability after discount:", probability)
```
